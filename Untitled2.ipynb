```python
import numpy as np
import matplotlib.pyplot as plt
```

## 1. Tsiolkovsky Rocket Equation

The fundamental equation of rocket motion:

**ΔV = Isp × g0 × ln(m0 / mk)**

Where:
- **ΔV** — delta-V, characteristic velocity change (m/s)
- **Isp** — specific impulse (seconds)
- **g0** — standard gravity, 9.81 m/s²
- **m0** — initial mass (kg)
- **mk** — final mass (kg)

```python
def delta_v(Isp, m0, mk):
  g = 9.81
  return Isp * g * np.log(m0 / mk)
```

```python
def fuel_mass(m0, mk):
  return m0 - mk
```

```python
def fuel_for_dv(dv_required, Isp, m0):
  g = 9.81
  mk = m0 / np. exp(dv_required / (Isp * g))
  return m0 - mk
```

## 2. Results

Comparison of chemical and electric propulsion for a spacecraft with
initial mass m0 = 1000 kg.

```python
Isp_chem = 300
Isp_ep = 2500
m0 = 1000
mk = 500

dv_chem = delta_v(Isp_chem, m0, mk)
dv_ep = delta_v(Isp_ep, m0, mk)

print("Chemical propulsion delta-V:", round(dv_chem, 1), "m/s")
print("Electric propulsion delta-V:", round(dv_ep, 1), "m/s")
print("Ratio:", round(dv_ep / dv_chem, 1))
```

```python
dv_mission = 4200  # m/s, LEO to GTO transfer

fuel_chem = fuel_for_dv(dv_mission, Isp_chem, m0)
fuel_ep = fuel_for_dv(dv_mission, Isp_ep, m0)

print("Fuel required (chemical):", round(fuel_chem, 1), "kg")
print("Fuel required (electric):", round(fuel_ep, 1), "kg")
print("Fuel saved with EP:", round(fuel_chem - fuel_ep, 1), "kg")
```

```python
mk_values = np.linspace(100, 900, 50)

dv_chem_values = delta_v(Isp_chem, m0, mk_values)
dv_ep_values = delta_v(Isp_ep, m0, mk_values)

plt.figure(figsize=(8, 5))
plt.plot(mk_values, dv_chem_values, label='Chemical (Isp=300 s)', color='blue', linewidth=2)
plt.plot(mk_values, dv_ep_values, label='Electric (Isp=2500 s)', color='green', linewidth=2)
plt.grid(True)
plt.xlabel('Final mass, kg')
plt.ylabel('Delta-V, m/s')
plt.title('Delta-V vs Final Mass')
plt.legend()
plt.show()
```

```python
dv_range = np.linspace(1000, 8000, 50)

fuel_chem_range = fuel_for_dv(dv_range, Isp_chem, m0)
fuel_ep_range = fuel_for_dv(dv_range, Isp_ep, m0)

plt.figure(figsize=(8, 5))
plt.plot(dv_range, fuel_chem_range, label='Chemical (Isp=300 s)', color='red', linewidth=2)
plt.plot(dv_range, fuel_ep_range, label='Electric (Isp=2500 s)', color='purple', linewidth=2)
plt.grid(True)
plt.xlabel('Required delta-V, m/s')
plt.ylabel('Fuel mass, kg')
plt.title('Fuel Mass vs Required Delta-V')
plt.legend()
plt.show()
```

## 3. Conclusion

For a 4200 m/s transfer mission with a 1000 kg spacecraft:

- **Chemical propulsion** requires significantly more fuel due to lower Isp.
- **Electric propulsion** reduces fuel mass by a large factor, but provides lower thrust.

This demonstrates why electric propulsion is preferred for deep-space
and long-duration missions, while chemical propulsion remains necessary
for launch and rapid maneuvers.
